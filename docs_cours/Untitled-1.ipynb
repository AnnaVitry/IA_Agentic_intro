# Exercice 1 Stateless

Création d’une fonction “chatbot()” dans un script python dans lequel on utilisera une boucle while pour communiquer (user_input = input("\nClient : ")).
Si l’utilisateur utilise les mots exit, stop ou fin la conversation s'arrête. On communique avec ollama en utilisant la même fonction qu’avant (ollama.generate).


```python
def chatbot(model_name: str = "llama3.2:3b"):
    print(f"Chatbot Ollama prêt (modèle : {model_name})")
    print("Tapez 'exit', 'stop' ou 'fin' pour quitter.")

    while True:
        try:
            # Lecture de l'entrée utilisateur
            user_input = input("\nClient : ").strip()

            # Vérification des mots-clés de sortie
            if user_input.lower() in ("exit", "stop", "fin"):
                print("Chatbot : Au revoir !")
                break

            if not user_input:
                print("Veuillez entrer un texte.")
                continue

            # Envoi au modèle Ollama
            response = ollama.generate(model=model_name, prompt=user_input)

            # Affichage de la réponse
            print(f"Chatbot : {response.get('response', '').strip()}")

        except KeyboardInterrupt:
            print("\nInterruption par l'utilisateur.")
            break
        except Exception as e:
            print(f"Erreur : {e}")
            break


if __name__ == "__main__":
    # Lancer le chatbot
    chatbot()
```

# Exercice 2 Stateful

On va maintenant créer une deuxième fonction “chatbot_memory()” qui aura les mêmes critères d’arrêt que précédemment. La mémoire va se conserver de message en message en empilant les messages dans une liste de dictionnaire. Le message initial est:


```python
def chatbotmemory(model_name: str = "llama3.2:3b"):
    messages = [
        {
            "role": "system",
            "content": "Tu es un assistant technique professionnel. Tu dois aider le client en te basant sur les informations fournies dans la discussion.",
        }
    ]
    print(f"Chatbot Ollama prêt (modèle : {model_name})")
    print("Tapez 'exit', 'stop' ou 'fin' pour quitter.")

    while True:
        try:
            # Lecture de l'entrée utilisateur
            user_input = input("\nClient : ").strip()

            # Vérification des mots-clés de sortie
            if user_input.lower() in ("exit", "stop", "fin"):
                print("Chatbot : Au revoir !")
                break

            if not user_input:
                print("Veuillez entrer un texte.")
                continue

            # Ajout du message utilisateur à l'historique
            messages.append({"role": "user", "content": user_input})

            # Envoi de l'historique complet au modèle
            response = ollama.chat(model=model_name, messages=messages)

            # Récupération et affichage de la réponse
            bot_reply = response["message"]["content"].strip()
            print(f"Chatbot : {bot_reply}")

            # Ajout de la réponse du bot à l'historique
            messages.append({"role": "assistant", "content": bot_reply})

        except KeyboardInterrupt:
            print("\nInterruption par l'utilisateur.")
            break
        except Exception as e:
            print(f"Erreur : {e}")
            break


if __name__ == "__main__":
    # Lancer le chatbot
    chatbotmemory()
```

```python
"""
Chatbot LangChain LCEL (LangChain Expression Language) avec Ollama
- Utilisation de ChatPromptTemplate, MessagesPlaceholder
- Chaînage avec l'opérateur "|"
- Stateful (avec mémoire)
"""

from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import AIMessage, HumanMessage

# Modèle Ollama (assurez-vous qu'il est installé via `ollama pull llama3.2:3b`)
llm = ChatOllama(model="llama3.2:3b", temperature=0)

# Prompt avec placeholder pour l'historique
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "Tu es un assistant technique pro. Réponds de manière concise."),
        MessagesPlaceholder(variable_name="chat_history"),
        ("human", "{input}"),
    ]
)

# Chaîne LCEL : le prompt passe dans le LLM
chain = prompt | llm

# Historique des échanges (liste d'objets HumanMessage / AIMessage)
history = []

# Mots-clés pour arrêter la conversation
STOP_WORDS = {"exit", "stop", "fin"}


def chatbot_lcel():
    """
    Chatbot avec LCEL et gestion de la mémoire manuelle.
    """
    print("\n=== Chatbot LangChain LCEL avec mémoire ===")
    while True:
        user_input = input("\nClient : ").strip()
        if not user_input:
            print("⚠️ Veuillez entrer un texte.")
            continue

        if user_input.lower() in STOP_WORDS:
            print("👋 Fin de la conversation.")
            break

        try:
            # Appel du LLM via la chaîne LCEL
            response = chain.invoke({"input": user_input, "chat_history": history})

            # Récupération du texte de réponse
            reply_text = response.content.strip()
            print(f"Assistant : {reply_text}")

            # Mise à jour de l'historique
            history.append(HumanMessage(content=user_input))
            history.append(AIMessage(content=reply_text))

        except Exception as e:
            print(f"❌ Erreur : {e}")


if __name__ == "__main__":
    chatbot_lcel()
```

```python
"""
Exercices : Stratégies de Mémoire (Juin 2026)
1. Sliding Window (Oubli après k messages)
2 & 3. Moving Summary / Summary Buffer (Conservation de l'info via LangGraph)
"""

from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import (
    trim_messages,
    SystemMessage,
    HumanMessage,
    RemoveMessage,
)
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.runnables import RunnablePassthrough
from langchain_core.chat_history import InMemoryChatMessageHistory

# Imports modernes pour la gestion d'état complexe (LangGraph)
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from langgraph.checkpoint.memory import MemorySaver
from typing import Annotated, TypedDict
import operator  # NOUVEAU : Nécessaire pour notre accumulateur de tokens

# === 1. Configuration du modèle ===
llm = ChatOllama(model="llama3.2:3b", temperature=0)

# =====================================================================
# EXERCICE 0 : FULL BUFFER (La mémoire absolue)
# Conserve TOUT. Précision 100%, mais coût en tokens croissant rapidement.
# =====================================================================


def build_full_buffer_chain():
    """Construit une chaîne LCEL sans aucun filtre (Full Buffer)."""
    prompt = ChatPromptTemplate.from_messages(
        [
            ("system", "Tu es un assistant technique. Réponds de manière concise."),
            MessagesPlaceholder(variable_name="chat_history"),
            ("human", "{input}"),
        ]
    )

    # 💡 La différence est ici : Pas de 'trimmer'.
    # La chaîne prend l'historique brut tel quel.
    chain = prompt | llm

    store = {}

    def get_session_history(session_id: str):
        if session_id not in store:
            store[session_id] = InMemoryChatMessageHistory()
        return store[session_id]

    return RunnableWithMessageHistory(
        chain,
        get_session_history,
        input_messages_key="input",
        history_messages_key="chat_history",
    )


def exo_0_full_buffer():
    print("\n" + "=" * 50)
    print("🐘 EXERCICE 0 : FULL BUFFER (Mémoire d'éléphant)")
    print("=" * 50)

    chain = build_full_buffer_chain()
    config = {"configurable": {"session_id": "session_full"}}
    input_tokens = 0  # FOCUS SUR LES TOKENS EN ENTRÉE (Contexte)

    # Message 1 : L'information clé à mémoriser
    print("\n[User]: Mon plat préféré est le Rougail Saucisse. Mémorise ça !")
    res = chain.invoke(
        {"input": "Mon plat préféré est le Rougail Saucisse. Mémorise ça !"},
        config=config,
    )

    usage = res.usage_metadata.get("input_tokens", 0) if res.usage_metadata else 0
    input_tokens += usage
    print(f"[Bot]: {res.content} (Input tokens: {usage})")

    # Messages 2, 3, 4 : On tente de saturer le contexte (mais le Full Buffer garde tout)
    for i in range(2, 5):
        print(f"\n[User]: Faisons diversion. Quelle est la météo ? (Diversion {i - 1})")
        res = chain.invoke(
            {"input": f"Faisons diversion. Quelle est la météo ? (Diversion {i - 1})"},
            config=config,
        )
        usage = res.usage_metadata.get("input_tokens", 0) if res.usage_metadata else 0
        input_tokens += usage
        print(f"[Bot]: {res.content} (Input tokens: {usage})")

    # Message 5 : Test de mémorisation
    print("\n[User]: Quel est mon plat préféré ?")
    res = chain.invoke({"input": "Quel est mon plat préféré ?"}, config=config)
    usage = res.usage_metadata.get("input_tokens", 0) if res.usage_metadata else 0
    input_tokens += usage
    print(f"[Bot]: {res.content} (Input tokens: {usage})")

    print(
        "\n-> RÉSULTAT : Le modèle s'en souvient parfaitement (aucune perte de données)."
    )
    print(f"💰 Poids total du contexte (Full Buffer) : {input_tokens} input tokens")


# =====================================================================
# EXERCICE 1 : SLIDING WINDOW (La méthode LCEL)
# Conserve les k derniers messages en clair. Amnésie garantie pour le reste.
# =====================================================================


def build_sliding_window_chain(k=4):
    """Construit une chaîne LCEL avec une fenêtre glissante stricte."""
    prompt = ChatPromptTemplate.from_messages(
        [
            ("system", "Tu es un assistant technique. Réponds de manière concise."),
            MessagesPlaceholder(variable_name="chat_history"),
            ("human", "{input}"),
        ]
    )

    trimmer = trim_messages(
        max_tokens=k,  # On compte en messages pour l'exercice
        strategy="last",
        token_counter=len,
        include_system=True,
        allow_partial=False,
        start_on="human",
    )

    chain = (
        RunnablePassthrough.assign(
            chat_history=lambda x: trimmer.invoke(x["chat_history"])
        )
        | prompt
        | llm
    )

    store = {}

    def get_session_history(session_id: str):
        if session_id not in store:
            store[session_id] = InMemoryChatMessageHistory()
        return store[session_id]

    return RunnableWithMessageHistory(
        chain,
        get_session_history,
        input_messages_key="input",
        history_messages_key="chat_history",
    )


def exo_1_sliding_window():
    print("\n" + "=" * 50)
    print("🚀 EXERCICE 1 : SLIDING WINDOW (k=4 messages)")
    print("=" * 50)

    chain = build_sliding_window_chain(k=4)
    config = {"configurable": {"session_id": "session_sliding"}}
    input_tokens = 0

    # Message 1 : L'information clé à mémoriser
    print("\n[User]: Mon plat préféré est le Rougail Saucisse. Mémorise ça !")
    res = chain.invoke(
        {"input": "Mon plat préféré est le Rougail Saucisse. Mémorise ça !"},
        config=config,
    )

    usage = res.usage_metadata.get("input_tokens", 0) if res.usage_metadata else 0
    input_tokens += usage
    print(f"[Bot]: {res.content} (Input tokens: {usage})")

    # Messages 2, 3, 4 : On sature la fenêtre de contexte (k=4)
    for i in range(2, 5):
        print(f"\n[User]: Faisons diversion. Quelle est la météo ? (Diversion {i - 1})")
        res = chain.invoke(
            {"input": f"Faisons diversion. Quelle est la météo ? (Diversion {i - 1})"},
            config=config,
        )
        usage = res.usage_metadata.get("input_tokens", 0) if res.usage_metadata else 0
        input_tokens += usage
        print(f"[Bot]: {res.content} (Input tokens: {usage})")

    # Message 5 : Test de mémorisation
    print("\n[User]: Quel est mon plat préféré ?")
    res = chain.invoke({"input": "Quel est mon plat préféré ?"}, config=config)
    usage = res.usage_metadata.get("input_tokens", 0) if res.usage_metadata else 0
    input_tokens += usage
    print(f"[Bot]: {res.content} (Input tokens: {usage})")

    print("\n-> RÉSULTAT : Le modèle a oublié (Amnésie due à la fenêtre glissante).")
    print(f"📉 Poids total du contexte (Sliding Window) : {input_tokens} input tokens")


# =====================================================================
# ÉTATS PARTAGÉS POUR LES MÉMOIRES AVEC RÉSUMÉ (LangGraph)
# =====================================================================
class SummaryState(TypedDict):
    summary: str
    messages: Annotated[list, add_messages]
    input_tokens: Annotated[int, operator.add]


def run_graph_and_get_response(graph, user_input, config):
    """Helper 2026 : Utilise le streaming pour intercepter les messages avant leur suppression."""
    inputs = {"messages": [HumanMessage(content=user_input)]}
    # On écoute les mises à jour nœud par nœud (updates)
    for event in graph.stream(inputs, config, stream_mode="updates"):
        # Si le nœud 'conversation' vient de s'exécuter, on intercepte la réponse de l'IA
        if "conversation" in event:
            ai_message = event["conversation"]["messages"][0].content
            tokens = event["conversation"].get("input_tokens", 0)
            print(f"[Bot]: {ai_message} (Input tokens du nœud: {tokens})")
    # On retourne l'état final pour les statistiques globales
    return graph.get_state(config).values


# =====================================================================
# EXERCICE 2 : PURE SUMMARY BUFFER
# Résume TOUT le passé à chaque itération. Ne garde aucun message en clair.
# =====================================================================


def build_pure_summary_graph():
    def call_model(state: SummaryState):
        summary = state.get("summary", "")
        sys_msg = "Tu es un assistant pro."
        if summary:
            sys_msg += f"\n\nVoici le résumé de la conversation : {summary}"

        current_input = state["messages"][-1]
        response = llm.invoke([SystemMessage(content=sys_msg), current_input])
        tokens = (
            response.usage_metadata.get("input_tokens", 0)
            if response.usage_metadata
            else 0
        )
        return {"messages": [response], "input_tokens": tokens}

    def summarize_all(state: SummaryState):
        summary = state.get("summary", "")
        recent_exchange = state["messages"][-2:]

        prompt = (
            f"Mets à jour ce résumé avec le nouvel échange.\n"
            f"Ancien résumé : {summary}\n\n"
            f"Nouvel échange :\nHumain: {recent_exchange[0].content}\nAssistant: {recent_exchange[1].content}"
        )
        res = llm.invoke([HumanMessage(content=prompt)])
        tokens = res.usage_metadata.get("input_tokens", 0) if res.usage_metadata else 0

        delete_msgs = [RemoveMessage(id=m.id) for m in recent_exchange]
        return {"summary": res.content, "messages": delete_msgs, "input_tokens": tokens}

    workflow = StateGraph(SummaryState)
    workflow.add_node("conversation", call_model)
    workflow.add_node("summarize", summarize_all)

    workflow.add_edge(START, "conversation")
    workflow.add_edge("conversation", "summarize")
    workflow.add_edge("summarize", END)

    return workflow.compile(checkpointer=MemorySaver())


def exo_2_pure_summary():
    print("\n\n" + "=" * 50)
    print("📝 EXERCICE 2 : PURE SUMMARY BUFFER (Mémoire synthétique)")
    print("=" * 50)

    graph = build_pure_summary_graph()
    config = {"configurable": {"thread_id": "thread_pure_summary"}}

    print("\n[User]: Mon plat préféré est le Rougail Saucisse. Mémorise ça !")
    run_graph_and_get_response(
        graph, "Mon plat préféré est le Rougail Saucisse. Mémorise ça !", config
    )

    for i in range(2, 5):
        prompt = (
            f"Faisons diversion. Parle-moi de la lune en 1 phrase. (Diversion {i - 1})"
        )
        print(f"\n[User]: {prompt}")
        state = run_graph_and_get_response(graph, prompt, config)
        print(f"  [DEBUG] Résumé actuel : {state.get('summary', '')[:60]}...")
        print(
            f"  [DEBUG] Messages passés stockés en clair : {len(state.get('messages', []))} (Doit être 0 !)"
        )

    print("\n[User]: Quel est mon plat préféré ?")
    state = run_graph_and_get_response(graph, "Quel est mon plat préféré ?", config)

    print(
        f"\n-> RÉSULTAT : L'information survit, mais uniquement sous forme condensée."
    )
    print(
        f"📊 Poids total du contexte (Pure Summary) : {state.get('input_tokens', 0)} input tokens"
    )


# =====================================================================
# EXERCICE 3 : MOVING SUMMARY BUFFER (Le meilleur des deux mondes)
# Conserve les k derniers messages, ET résume les plus anciens.
# =====================================================================


def build_moving_summary_graph(k=4):
    def call_model(state: SummaryState):
        summary = state.get("summary", "")
        sys_msg = "Tu es un assistant pro."
        if summary:
            sys_msg += (
                f"\n\nVoici le résumé de la conversation jusqu'à présent : {summary}"
            )

        messages_to_send = [SystemMessage(content=sys_msg)] + state["messages"]
        response = llm.invoke(messages_to_send)
        tokens = (
            response.usage_metadata.get("input_tokens", 0)
            if response.usage_metadata
            else 0
        )
        return {"messages": [response], "input_tokens": tokens}

    def should_summarize(state: SummaryState):
        if len(state["messages"]) > k + 2:
            return "summarize_conversation"
        return END

    def summarize_conversation(state: SummaryState):
        summary = state.get("summary", "")
        messages_to_summarize = state["messages"][:-4]

        summary_prompt = (
            f"Résume la conversation ci-dessous de manière concise.\n"
            f"Résumé existant : {summary}\n\n"
            f"Nouveaux messages à inclure :\n"
            f"{[m.content for m in messages_to_summarize]}"
        )

        res = llm.invoke([HumanMessage(content=summary_prompt)])
        tokens = res.usage_metadata.get("input_tokens", 0) if res.usage_metadata else 0

        delete_messages = [RemoveMessage(id=m.id) for m in messages_to_summarize]
        return {
            "summary": res.content,
            "messages": delete_messages,
            "input_tokens": tokens,
        }

    workflow = StateGraph(SummaryState)
    workflow.add_node("conversation", call_model)
    workflow.add_node("summarize_conversation", summarize_conversation)

    workflow.add_edge(START, "conversation")
    workflow.add_conditional_edges("conversation", should_summarize)
    workflow.add_edge("summarize_conversation", END)

    return workflow.compile(checkpointer=MemorySaver())


def exo_3_moving_summary():
    print("\n\n" + "=" * 50)
    print("🧠 EXERCICE 3 : MOVING SUMMARY (LangGraph)")
    print("=" * 50)

    graph = build_moving_summary_graph(k=4)
    config = {"configurable": {"thread_id": "thread_moving_summary"}}

    print("\n[User]: Mon plat préféré est le Rougail Saucisse. Mémorise ça !")
    run_graph_and_get_response(
        graph, "Mon plat préféré est le Rougail Saucisse. Mémorise ça !", config
    )

    for i in range(2, 5):
        prompt = (
            f"Faisons diversion. Parle-moi de la lune en 1 phrase. (Diversion {i - 1})"
        )
        print(f"\n[User]: {prompt}")
        state = run_graph_and_get_response(graph, prompt, config)
        print(f"  [DEBUG] État du Résumé : {state.get('summary', 'Aucun')[:60]}...")
        print(
            f"  [DEBUG] Nombre de messages bruts en mémoire : {len(state.get('messages', []))}"
        )

    print("\n[User]: Quel est mon plat préféré ?")
    state = run_graph_and_get_response(graph, "Quel est mon plat préféré ?", config)

    print(
        "\n-> RÉSULTAT : Le modèle s'en souvient ! L'information a survécu grâce au résumé en toile de fond."
    )
    print(
        f"\n📈 Poids total du contexte (Moving Summary) : {state.get('input_tokens', 0)} input tokens"
    )


if __name__ == "__main__":
    exo_0_full_buffer()  # 🐘 Le Full Buffer (Précision 100%, Prix maximum)
    exo_1_sliding_window()  # 🚀 La Sliding Window (Amnésique mais pas chère)
    exo_2_pure_summary()  # 📝 Le Summary Buffer (Très synthétique, coût constant)
    exo_3_moving_summary()  # 🧠 Le Moving Summary (L'équilibre parfait)
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage


def chatbot_avec_memoire_profonde():
    """
    Ce chatbot combine deux concepts :
    1. Une mémoire à court terme (history_session) pour la fluidité immédiate.
    2. Une mémoire à long terme (ChromaDB) pour les souvenirs lointains.
    """
    print("\n" + "=" * 50)
    print("🧠 CHATBOT AVEC MÉMOIRE PROFONDE (ChromaDB)")
    print("=" * 50)
    print("Tapez 'exit', 'stop' ou 'fin' pour quitter.")
    print(
        "💡 Astuce : Donnez lui votre plat préféré, quittez le script, puis relancez-le pour tester son souvenir !"
    )

    # 1. On définit la personnalité et la structure du prompt.
    # Notez la variable {archives} qui recevra les résultats de la recherche vectorielle.
    prompt = ChatPromptTemplate.from_messages(
        [
            (
                "system",
                """Tu es un assistant technique ultra-compétent. 
Tu disposes d'archives d'anciennes conversations avec cet utilisateur.
Utilise ces archives pour personnaliser tes réponses si elles sont pertinentes.

--- DÉBUT DES ARCHIVES PERTINENTES ---
{archives}
--- FIN DES ARCHIVES ---

Réponds toujours de manière polie et concise. Si l'information n'est ni dans ta mémoire 
immédiate ni dans les archives, dis-le honnêtement.""",
            ),
            MessagesPlaceholder(variable_name="session_history"),
            ("human", "{input}"),
        ]
    )

    # 2. On crée notre chaîne LCEL (Prompt -> Modèle)
    chain = prompt | llm

    # Historique court terme (mémoire vive de la session en cours)
    history_session = []

    while True:
        user_input = input("\nClient : ").strip()

        if not user_input:
            continue

        if user_input.lower() in ("exit", "stop", "fin"):
            print("👋 Au revoir ! Vos souvenirs sont scellés dans ChromaDB.")
            break

        try:
            # --- ÉTAPE A : RECHERCHE DANS LE PASSÉ (RETRIEVAL) ---
            # On cherche les 3 morceaux d'historique les plus proches mathématiquement
            # de la question que l'utilisateur vient de poser.
            docs_du_passe = vectorstore.similarity_search(user_input, k=3)

            # On formate ces morceaux en un seul texte lisible par le LLM
            context_archives = "\n".join([doc.page_content for doc in docs_du_passe])

            # --- ÉTAPE B : GÉNÉRATION DE LA RÉPONSE (AUGMENTATION & GENERATION) ---
            response = chain.invoke(
                {
                    "input": user_input,
                    "session_history": history_session,
                    "archives": context_archives
                    if context_archives
                    else "Aucune archive trouvée.",
                }
            )

            bot_reply = response.content.strip()
            print(f"Assistant : {bot_reply}")

            # --- ÉTAPE C : SAUVEGARDE POUR LE FUTUR ---
            # 1. Mise à jour de la mémoire à court terme (pour la suite immédiate de la conversation)
            history_session.append(HumanMessage(content=user_input))
            history_session.append(AIMessage(content=bot_reply))

            # 2. Mise à jour de la mémoire à long terme (ChromaDB)
            # On formate l'échange sous forme d'une chaîne de caractères explicite
            echange_formate = f"L'utilisateur a dit: '{user_input}' | L'assistant a répondu: '{bot_reply}'"

            # On l'ajoute à la base de données. ChromaDB va automatiquement calculer
            # l'embedding en arrière-plan et le stocker sur le disque.
            vectorstore.add_texts(texts=[echange_formate])

        except Exception as e:
            print(f"❌ Une erreur est survenue : {e}")


if __name__ == "__main__":
    chatbot_avec_memoire_profonde()
```

```python
from langchain_community.vectorstores import PGVector
from langchain_ollama import OllamaEmbeddings
from langchain_core.documents import Document


def demo_pgvector():
    print("=" * 50)
    print("🐘 DÉMONSTRATION PGVECTOR (Recherche Hybride)")
    print("=" * 50)

    # Initialisation de notre modèle d'embedding local
    print("🧠 Chargement du modèle d'embedding...")
    embeddings = OllamaEmbeddings(model="llama3.2:3b")

    # Chaîne de connexion PostgreSQL (Syntaxe SQLAlchemy)
    # Format : postgresql+driver://utilisateur:mot_de_passe@hote:port/nom_de_la_base
    CONNECTION_STRING = (
        "postgresql+psycopg2://postgres:password@localhost:5432/postgres"
    )

    print("🔌 Connexion à PostgreSQL...")
    try:
        # Initialisation du VectorStore
        # LangChain va automatiquement créer les tables nécessaires dans Postgres si elles n'existent pas
        vectorstore = PGVector(
            connection_string=CONNECTION_STRING,
            collection_name="historique_assistant",
            embedding_function=embeddings,
            use_jsonb=True,  # Optimisation vitale pour PostgreSQL (stockage des métadonnées au format JSON binaire)
        )
        print("✅ Connecté à PostgreSQL avec succès !")
    except Exception as e:
        print("\n❌ ERREUR DE CONNEXION :")
        print("Avez-vous bien lancé un serveur PostgreSQL avec l'extension pgvector ?")
        print(f"Détail technique : {e}")
        return

    # =====================================================================
    # LA PUISSANCE DE PGVECTOR : L'ajout de Métadonnées relationnelles
    # =====================================================================
    print(
        "\n📝 Ajout de documents avec des métadonnées (ID utilisateur, date, type)..."
    )

    # Au lieu d'ajouter du texte brut, on ajoute des Objets "Document"
    # pour y attacher des dictionnaires de métadonnées.
    documents_a_sauvegarder = [
        Document(
            page_content="Facture n°1204 : Achat de 3 serveurs cloud pour 4500€.",
            metadata={"user_id": 42, "type": "facture", "date": "2026-05-01"},
        ),
        Document(
            page_content="Rapport de bug : Le bouton de connexion crash sur Safari.",
            metadata={"user_id": 42, "type": "ticket_tech", "date": "2026-06-01"},
        ),
        Document(
            page_content="Facture n°1205 : Renouvellement domaine web pour 15€.",
            metadata={"user_id": 99, "type": "facture", "date": "2026-06-02"},
        ),
    ]

    # Ajout dans la base Postgres (calcul des vecteurs + insertion SQL)
    vectorstore.add_documents(documents_a_sauvegarder)
    print("✅ Documents insérés dans la table SQL !")

    # =====================================================================
    # RECHERCHE HYBRIDE : Vecteur Sémantique + Filtre SQL (WHERE)
    # =====================================================================
    query = "Je cherche des informations sur des paiements ou des dépenses."

    print(f"\n🔍 Recherche sémantique pour : '{query}'")
    print("Filtre SQL appliqué : UNIQUEMENT les documents du user_id 42")

    # On utilise un filtre (filter={...}) qui va se traduire par une clause WHERE en SQL
    # TOUT en faisant un calcul de similarité cosinus (ORDER BY distance)
    resultats = vectorstore.similarity_search(
        query=query,
        k=2,
        filter={"user_id": 42},  # Le filtre magique rendu possible par Postgres !
    )

    print("\n--- RÉSULTATS TROUVÉS ---")
    for i, doc in enumerate(resultats, 1):
        print(f"Résultat {i} :")
        print(f"  Texte : {doc.page_content}")
        print(f"  Métadonnées : {doc.metadata}")

    print(
        "\n💡 Remarquez comment la facture du user_id 99 a été ignorée grâce au filtre SQL, même si sémantiquement elle parlait aussi d'une dépense !"
    )


if __name__ == "__main__":
    demo_pgvector()
```

# RAG


## Exercice

Vous allez faire un service de chatbot qui renseignera les clients sur le contenu du menu.
Le fichier menu.txt vous sera fourni et le chat bot devra se comporter comme un serveur virtuel, s’il ne connaît pas la réponse il doit rester poli.
En utilisant LangChain et la structure LCEL (les pipes |), développez un script qui :
Charge le fichier menu.txt.
Découpe le texte de manière intelligente (utilisez RecursiveCharacterTextSplitter).
Stocke les informations dans une base vectorielle ChromaDB.
Interroge le modèle Llama 3.2 via Ollama en lui fournissant les morceaux du menu les plus pertinents (k=3).
Jouez avec les paramètres, changez les méthodes de chunks. Notez tout ce que vous trouvez anormal.


```python
import os
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_ollama import ChatOllama, OllamaEmbeddings
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser


def chatbot_restaurant_rag():
    print("=" * 50)
    print("CHATBOT RAG : SERVEUR DU GLITCH GOURMAND")
    print("=" * 50)

    # 0. Préparation du terrain
    model_name = "llama3.2:3b"
    llm = ChatOllama(
        model=model_name, temperature=0
    )  # Température 0 : on veut des faits, pas de la poésie
    embeddings = OllamaEmbeddings(model=model_name)

    print("1. INGESTION : Lecture du document...")
    loader = TextLoader("./menu.txt", encoding="utf-8")
    documents = loader.load()

    print("2. CHUNKING : Découpage intelligent du texte...")
    # RecursiveCharacterTextSplitter est la norme de l'industrie. Il essaie
    # de couper aux doubles sauts de lignes d'abord, puis aux sauts simples, etc.
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=300,  # Petite taille de chunk adaptée à un menu
        chunk_overlap=50,  # Recouvrement pour ne pas couper le prix du plat !
        separators=["\n\n", "\n", " ", ""],
    )
    chunks = text_splitter.split_documents(documents)
    print(f"   -> Le menu a été découpé en {len(chunks)} morceaux (chunks).")

    print("3 & 4. EMBEDDING & RETRIEVAL : Indexation dans ChromaDB...")
    # Base vectorielle éphémère (en mémoire) pour l'exercice
    vectorstore = Chroma.from_documents(chunks, embedding=embeddings)

    # Le retriever ira chercher les 3 (k=3) morceaux les plus pertinents
    retriever = vectorstore.as_retriever(search_kwargs={"k": 3})

    print("5. AUGMENTATION : Création de la chaîne (Pipe LCEL)...")

    # Le prompt système dicte la personnalité stricte du RAG (Anti-hallucination)
    template = """Tu es le serveur virtuel du restaurant 'Le GLITCH GOURMAND'. 
Réponds à la question du client en utilisant UNIQUEMENT les données du menu fournies ci-dessous.
Si la réponse ne s'y trouve pas, réponds poliment que tu ne sais pas ou que ce n'est pas à la carte.
Ne mentionne jamais le mot "contexte" ou "données fournies" au client.

Menu (Contexte extrait par RAG):
{context}

Question client: {question}
"""
    prompt = ChatPromptTemplate.from_template(template)

    # Fonction pour assembler les chunks trouvés en un seul texte
    def format_docs(docs):
        return "\n\n".join([d.page_content for d in docs])

    # Le cœur du RAG en LCEL (LangChain Expression Language) :
    # 1. On prend le dictionnaire d'entrée.
    #    La clé 'context' se remplit en cherchant dans la BDD (retriever) puis en formatant.
    #    La clé 'question' est récupérée telle quelle grâce à RunnablePassthrough().
    # 2. On injecte tout ça dans le Prompt.
    # 3. On envoie au LLM.
    # 4. On extrait la réponse pure (StrOutputParser).
    rag_chain = (
        {"context": retriever | format_docs, "question": RunnablePassthrough()}
        | prompt
        | llm
        | StrOutputParser()
    )

    print("✅ Le serveur est prêt à prendre votre commande !")
    print(
        "Essayez de lui demander s'il y a des options végétariennes, le prix du burger, ou s'ils servent des pizzas.\n"
    )

    # Boucle de discussion
    while True:
        question_test = input("Client : ").strip()
        if not question_test:
            continue
        if question_test.lower() in ("exit", "stop", "fin"):
            print("Serveur : Au revoir et à bientôt au GLITCH GOURMAND !")
            break

        try:
            # Invocation magique de notre pipeline LCEL
            response = rag_chain.invoke(question_test)
            print(f"Serveur : {response}\n")
        except Exception as e:
            print(f"Erreur en cuisine : {e}\n")


if __name__ == "__main__":
    chatbot_restaurant_rag()
```

```python
# =====================================================================
# ÉVOLUTION : Lecture d'un fichier utilisateur (menu.txt),
# Prétraitement des données (Data Cleaning) et Chunking Structurel.
# =====================================================================

import os
import re  # Bibliothèque pour les Expressions Régulières (Regex)
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import MarkdownHeaderTextSplitter
from langchain_chroma import Chroma
from langchain_ollama import ChatOllama, OllamaEmbeddings
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser


def preparer_rag(k=3):
    """
    Prépare la chaîne RAG avec une valeur k personnalisable.
    Lit le fichier menu.txt de l'utilisateur et le nettoie en mémoire.
    """
    model_name = "llama3.2:3b"
    llm = ChatOllama(model=model_name, temperature=0)
    embeddings = OllamaEmbeddings(model=model_name)

    # 1. VÉRIFICATION ET INGESTION DU FICHIER
    if not os.path.exists("menu.txt"):
        raise FileNotFoundError(
            "⚠️ Le fichier 'menu.txt' est introuvable. Placez-le dans le même dossier que le script."
        )

    with open("menu.txt", "r", encoding="utf-8") as f:
        doc_texte = f.read()

    # =====================================================================
    # 🌟 DATA CLEANING (Prétraitement) : Le secret d'un bon RAG !
    # Le fichier utilise des '--- JOUR ---' et 'Service :'.
    # On va les convertir en balises Markdown '# JOUR' et '## Service'
    # en mémoire, pour que notre Splitter les reconnaisse parfaitement.
    # =====================================================================

    # 1. Transforme "--- LUNDI ---" en "# LUNDI"
    doc_texte = re.sub(r"---\s*(.*?)\s*---", r"# \1", doc_texte)

    # 2. Transforme "Service du Midi :" en "## Service du Midi"
    doc_texte = re.sub(r"(Service.*?)\s*:", r"## \1", doc_texte)

    # 3. Transforme la section Infos pratiques
    doc_texte = doc_texte.replace("INFOS PRATIQUES :", "# INFOS PRATIQUES")

    # 2. CHUNKING STRUCTUREL
    # On dit au splitter de couper à chaque fois qu'il voit un '#' ou '##'
    headers_to_split_on = [
        ("#", "Jour"),  # Le '#' deviendra une métadonnée 'Jour'
        ("##", "Service"),  # Le '##' deviendra une métadonnée 'Service'
    ]
    markdown_splitter = MarkdownHeaderTextSplitter(
        headers_to_split_on=headers_to_split_on
    )

    # Découpage logique. Chaque chunk aura {metadata: {'Jour': 'LUNDI', 'Service': 'Service du Midi'}}
    chunks = markdown_splitter.split_text(doc_texte)

    # 3 & 4. EMBEDDING ET RETRIEVAL
    vectorstore = Chroma.from_documents(chunks, embedding=embeddings)
    retriever = vectorstore.as_retriever(search_kwargs={"k": k})

    template = """Tu es le serveur virtuel du restaurant 'Le Glitch Gourmand'. 
Réponds en utilisant UNIQUEMENT le menu ci-dessous.

Menu (Contexte extrait de la base de données):
{context}

Question client: {question}
"""
    prompt = ChatPromptTemplate.from_template(template)

    # On modifie le formateur pour inclure les métadonnées (Jour et Service)
    def format_docs(docs):
        text_context = ""
        for d in docs:
            jour = d.metadata.get("Jour", "Général")
            service = d.metadata.get("Service", "")
            # Construit un entête visible pour le LLM, ex: [MARDI - Service du Soir]
            entete = f"[{jour}" + (f" - {service}]" if service else "]")
            text_context += f"{entete} : {d.page_content}\n\n"
        return text_context

    rag_chain = (
        {"context": retriever | format_docs, "question": RunnablePassthrough()}
        | prompt
        | llm
        | StrOutputParser()
    )

    return rag_chain, retriever, format_docs


def test_automatise_k():
    """
    Fonction pour visualiser l'impact de 'k' avec votre propre menu.
    """
    print("\n" + "=" * 50)
    print("🧪 MODE TEST AUTOMATISÉ : L'impact du paramètre K")
    print("=" * 50)

    # Une question complexe qui demande de piocher à plusieurs endroits du menu
    question = "Je voudrais venir mercredi soir pour une pizza, et mon ami veut un plat végan lundi midi. Que proposez-vous ?"
    print(f"Question test : '{question}'\n")

    for k_test in [1, 4]:
        print(f"--- TEST AVEC k = {k_test} ---")
        _, retriever, formatter = preparer_rag(k=k_test)

        chunks_trouves = retriever.invoke(question)
        contexte_genere = formatter(chunks_trouves)

        print(
            f"Le LLM recevra ce contexte (longueur: {len(contexte_genere)} caractères) :"
        )
        print("-" * 20)
        print(contexte_genere.strip())
        print("-" * 20)
        print("💡 Analyse :")
        if k_test == 1:
            print(
                "Avec k=1, l'IA ne voit qu'un seul morceau du puzzle (soit lundi, soit mercredi). Elle sera incapable de répondre complètement."
            )
        elif k_test == 4:
            print(
                "Avec k=4, le contexte est plus large. Elle devrait trouver à la fois le lundi midi (Végan) et le mercredi soir (Pizza)."
            )
        print("\n")


def chatbot_restaurant_rag():
    print("=" * 50)
    print("💻 BIENVENUE CHEZ 'LE GLITCH GOURMAND'")
    print("=" * 50)

    choix = input("Voulez-vous lancer le test automatisé sur le paramètre K ? (o/n) : ")
    if choix.lower() == "o":
        test_automatise_k()

    print("\n🍽️ DÉMARRAGE DU SERVEUR (k=3)")
    rag_chain, _, _ = preparer_rag(k=3)

    print("✅ Le serveur est prêt !")

    while True:
        question_test = input("\nClient : ").strip()
        if not question_test:
            continue
        if question_test.lower() in ("exit", "stop", "fin"):
            print("Serveur : Au revoir, et attention aux bugs digestifs !")
            break

        try:
            response = rag_chain.invoke(question_test)
            print(f"Serveur : {response}")
        except Exception as e:
            print(f"Erreur serveur (500) : {e}\n")


if __name__ == "__main__":
    chatbot_restaurant_rag()
```

# =====================================================================
# Objectif : Créer un chatbot capable d'ingérer différents formats
# (PDF, TXT, MD) depuis un dossier, sans écrire de parseurs complexes,
# en utilisant FAISS pour une recherche ultra-rapide en mémoire RAM.
# =====================================================================

```python
import os
import faiss
from llama_index.core import (
    SimpleDirectoryReader,
    VectorStoreIndex,
    StorageContext,
    Settings,
)
from llama_index.vector_stores.faiss import FaissVectorStore
from llama_index.llms.ollama import Ollama
from llama_index.embeddings.ollama import OllamaEmbedding


def main():
    print("=" * 50)
    print("🦙 RAG MULTISUPPORT AVEC LLAMA INDEX & FAISS")
    print("=" * 50)

    # 1. Vérification du dossier contenant le référentiel Microsoft/Simplon
    data_dir = "./data_referentiel"
    if not os.path.exists(data_dir):
        os.makedirs(data_dir)
        print(f"⚠️ Le dossier '{data_dir}' n'existait pas. Je viens de le créer.")
        print(
            "Veuillez y glisser vos fichiers (IA-E1.pdf, IA-E2-E3.md, IA-E4-E5.txt) et relancer le script."
        )
        return

    if len(os.listdir(data_dir)) == 0:
        print(
            f"⚠️ Le dossier '{data_dir}' est vide. Mettez-y vos fichiers de référentiel !"
        )
        return

    print("\n⚙️ 1. Configuration des modèles locaux (Ollama)...")
    # Dans les versions modernes de Llama Index (>v0.10), on configure les modèles
    # de manière globale via l'objet 'Settings'.
    model_name = "llama3.2:3b"
    embed_model_name = "nomic-embed-text"  # <-- NOUVEAU : Un modèle ultra-léger dédié à 100% aux vecteurs !

    # On bride volontairement la fenêtre de contexte à 4096 tokens pour éviter
    # qu'Ollama ne réserve 10 Go de RAM inutilement pour son "KV Cache".
    Settings.llm = Ollama(
        model=model_name, temperature=0, request_timeout=120.0, context_window=4096
    )
    Settings.embed_model = OllamaEmbedding(model_name=embed_model_name)

    print("📂 2. Chargement automatique des documents (PDF, TXT, MD)...")
    # C'est la force de Llama Index : un seul objet gère tous les formats !
    documents = SimpleDirectoryReader(data_dir).load_data()
    print(f"   -> Succès : {len(documents)} pages/morceaux extraits.")

    print("⚡ 3. Initialisation de la base vectorielle FAISS (En RAM)...")
    # ⚠️ ATTENTION : La dimension dépend strictement de votre modèle d'Embedding !
    # Le modèle 'nomic-embed-text' génère des vecteurs de taille 768 (et non plus 3072 comme Llama 3.2).
    # Si on laisse 3072, FAISS plantera avec une erreur 'Dimension mismatch'.
    d = 768
    faiss_index = faiss.IndexFlatL2(d)

    # On encapsule l'index FAISS natif dans la structure Llama Index
    vector_store = FaissVectorStore(faiss_index=faiss_index)
    storage_context = StorageContext.from_defaults(vector_store=vector_store)

    print("🧠 4. Création de l'Index (Calcul des Embeddings)...")
    # Llama Index va "chunker" (découper) automatiquement les documents
    # et calculer les embeddings via Ollama avant de les stocker dans FAISS.
    index = VectorStoreIndex.from_documents(documents, storage_context=storage_context)

    print("✅ Indexation terminée !")

    # 5. Création du Query Engine (Le Moteur de Recherche / Chatbot)
    # Llama Index crée automatiquement le prompt optimal sous le capot.
    query_engine = index.as_query_engine(
        similarity_top_k=3  # On limite aux 3 meilleurs morceaux (chunks) pour ne pas noyer le LLM
    )

    print("\n" + "=" * 50)
    print("🎓 EXAMINATEUR DU RÉFÉRENTIEL MICROSOFT/SIMPLON PRÊT !")
    print("Posez vos questions sur les compétences E1, E2, E3, E4, E5.")
    print("Tapez 'exit', 'stop' ou 'fin' pour quitter.")
    print("=" * 50 + "\n")

    # Boucle de conversation
    while True:
        user_input = input("Candidat : ").strip()
        if not user_input:
            continue
        if user_input.lower() in ["exit", "stop", "fin"]:
            print("Examinateur : Fin de l'évaluation. Au revoir !")
            break

        try:
            print("⏳ Recherche dans le référentiel et réflexion en cours...")
            # Une seule ligne suffit : 'query' fait le retrieval ET la génération
            response = query_engine.query(user_input)

            print(f"\n🤖 Llama-Index : {response}\n")

            # --- ASTUCE DEBUG : Afficher les sources ---
            # Décommentez les lignes ci-dessous pour voir d'où le LLM a tiré sa réponse
            # print("   [Sources utilisées :]")
            # for source in response.source_nodes:
            #     fichier = source.metadata.get('file_name', 'Inconnu')
            #     print(f"   - Fichier : {fichier}")
            # ------------------------------------------

        except Exception as e:
            print(f"❌ Erreur lors de l'interrogation : {e}\n")


if __name__ == "__main__":
    main()
```

```python
# %% EXERCICE : L'Agent Autonome connecté à PostgreSQL (PGVector)
# =====================================================================
# Objectif : Créer un agent ReAct capable de fouiller de lui-même dans
# une base de données d'entreprise et d'utiliser une calculatrice.
# =====================================================================

import getpass
from typing import Annotated
from langchain_ollama import ChatOllama, OllamaEmbeddings
from langchain_core.messages import SystemMessage, HumanMessage
from langchain_core.tools import tool
from langchain_community.vectorstores import PGVector
from langchain_core.documents import Document
from langgraph.graph import StateGraph, START, END, MessagesState
from langgraph.prebuilt import ToolNode, tools_condition
from langgraph.checkpoint.memory import MemorySaver


def initialiser_base_de_donnees():
    """Gère la connexion à Postgres et l'insertion de quelques données factices."""
    print("\n1. Authentification PostgreSQL")
    db_user = input("Utilisateur Postgres (Défaut: postgres) : ") or "postgres"
    db_password = (
        getpass.getpass(f"Mot de passe pour '{db_user}' (Défaut: password) : ")
        or "password"
    )

    # Chaîne de connexion
    CONNECTION_STRING = (
        f"postgresql+psycopg2://{db_user}:{db_password}@localhost:5432/postgres"
    )
    embeddings = OllamaEmbeddings(model="nomic-embed-text")

    print("Connexion et initialisation de PGVector...")
    try:
        vectorstore = PGVector(
            connection_string=CONNECTION_STRING,
            collection_name="archives_comptables",
            embedding_function=embeddings,
            use_jsonb=True,
        )

        # On insère des données pour que l'agent ait quelque chose à trouver !
        docs = [
            Document(
                page_content="Facture n°101 : Achat de licences logicielles pour 1200 euros HT.",
                metadata={"type": "facture"},
            ),
            Document(
                page_content="Facture n°102 : Prestation de conseil Cloud pour 3500 euros HT.",
                metadata={"type": "facture"},
            ),
            Document(
                page_content="Note RH : L'employé Paul a pris 3 jours de congés.",
                metadata={"type": "rh"},
            ),
        ]
        vectorstore.add_documents(docs)
        print("Base de données prête et peuplée !")
        return vectorstore

    except Exception as e:
        print(f"\nERREUR DE CONNEXION POSTGRESQL : {e}")
        print("Vérifiez que votre serveur tourne et que le mot de passe est bon.")
        return None


def main():
    print("=" * 50)
    print("CRÉATION DE L'AGENT COMPTABLE (PGVector + LangGraph)")
    print("=" * 50)

    # 1. Initialisation de la BDD
    vectorstore = initialiser_base_de_donnees()
    if not vectorstore:
        return  # Arrêt d'urgence si la BDD est inaccessible

    # ---------------------------------------------------------
    # 2. DÉFINITION DES OUTILS DE L'AGENT
    # ---------------------------------------------------------
    print("\n2. Création des outils de l'Agent...")

    # OUTIL 1 : Recherche dans la base de données vectorielle
    @tool
    def rechercher_archives_entreprise(requete: str) -> str:
        """Recherche des documents internes, factures ou notes RH dans la base de données sécurisée de l'entreprise.
        L'argument 'requete' DOIT être une phrase précise décrivant ce que tu cherches (ex: 'prestation de conseil Cloud')."""
        print(f"      [BDD] Recherche vectorielle en cours pour : '{requete}'...")

        # SOLUTION 1 : On augmente 'k' à 3 pour récupérer plus de contexte et éviter les faux positifs
        resultats = vectorstore.similarity_search(requete, k=3)
        if not resultats:
            return "Aucun document trouvé dans la base de données."

        # On renvoie tous les résultats trouvés (séparés par un tiret) pour que l'Agent fasse le tri lui-même
        return "\n".join([f"- {doc.page_content}" for doc in resultats])

    # OUTIL 2 : La calculatrice de TVA
    @tool
    def calculateur_tva(prix_ht: float) -> float:
        """Calcule le prix TTC à partir du prix HT en ajoutant une TVA de 20%.
        Utile pour les calculs financiers précis."""
        print(f"      [CALC] Application de la TVA sur : {prix_ht}€...")
        return float(prix_ht) * 1.20

    tools = [rechercher_archives_entreprise, calculateur_tva]

    # ---------------------------------------------------------
    # 3. CONFIGURATION DU CERVEAU ET DU GRAPHE
    # ---------------------------------------------------------
    print("3. Assemblage du réseau neuronal (Llama 3.2 + Tools)...")
    llm = ChatOllama(model="llama3.2:3b", temperature=0)
    llm_with_tools = llm.bind_tools(tools)

    def node_agent(state: MessagesState):
        response = llm_with_tools.invoke(state["messages"])
        return {"messages": [response]}

    # Création du Graphe
    workflow = StateGraph(MessagesState)
    workflow.add_node("agent", node_agent)
    workflow.add_node("tools", ToolNode(tools))

    workflow.add_edge(START, "agent")
    workflow.add_conditional_edges("agent", tools_condition)
    workflow.add_edge("tools", "agent")

    # On n'oublie pas le MemorySaver !
    memory = MemorySaver()
    agent_app = workflow.compile(checkpointer=memory)

    # ---------------------------------------------------------
    # 4. INVOCATION : LE DOUBLE COMBO !
    # ---------------------------------------------------------
    print("\n" + "=" * 50)
    print("DÉMARRAGE DE L'AGENT COMPTABLE")
    print("=" * 50)

    # SOLUTION 2 : Un prompt système beaucoup plus directif pour encadrer les faiblesses du LLM
    system_prompt = SystemMessage(
        content="""Tu es l'assistant comptable de l'entreprise.
1. Si on te demande des informations sur une facture, UTILISE l'outil 'rechercher_archives_entreprise'. Fournis-lui de vrais mots-clés en argument.
2. Si tu dois donner un montant TTC, UTILISE le 'calculateur_tva'.
ATTENTION : Ne donne jamais tes propres connaissances. Base-toi uniquement sur les retours de tes outils.
Réponds de manière professionnelle en expliquant ta démarche."""
    )

    query = "Peux-tu fouiller dans nos archives pour trouver le montant de la prestation de conseil Cloud, et me dire combien cela fera TTC au total ?"

    print(f"Question du Boss : {query}\n")
    print("L'Agent au travail...\n")

    inputs = {"messages": [system_prompt, HumanMessage(content=query)]}
    config = {"configurable": {"thread_id": "compta_1"}}

    # On observe le cheminement de l'agent en direct
    for s in agent_app.stream(inputs, config, stream_mode="values"):
        message = s["messages"][-1]
        if message.type == "ai" and message.tool_calls:
            nom_outil = message.tool_calls[0]["name"]
            args_outil = message.tool_calls[0]["args"]
            print(
                f"   [DÉCISION] J'ai besoin de l'outil '{nom_outil}' avec les arguments : {args_outil}"
            )
        elif message.type == "tool":
            print(f"   [RETOUR OUTIL] {message.content}\n")

    final_state = agent_app.get_state(config).values
    print("\nRÉPONSE FINALE AU BOSS :")
    print(final_state["messages"][-1].content)


if __name__ == "__main__":
    main()
```
