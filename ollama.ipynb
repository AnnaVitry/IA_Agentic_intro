## CHATBOT


### Exercice 1 Stateless

Création d’une fonction “chatbot()” dans un script python dans lequel on utilisera une
boucle while pour communiquer (`user_input = input("\nClient : ")`). ​
Si l’utilisateur utilise les mots exit, stop ou fin la conversation s'arrête. On communique
avec ollama en utilisant la même fonction qu’avant (`ollama.generate`).


````python
def chatbot():
    while True:
        user_input = input("\nClient : ")
        if user_input.lower() in ["exist", "stop", "fin"]:
            break

        reponse = ollama.generate(model="llama3.2:3b", prompt=user_input)
        print("IA :", reponse["response"])


chatbot()
````

### Exercice 2 Stateful

On va maintenant créer une deuxième fonction “chatbot_memory()” qui aura les mêmes
critères d’arrêt que précédemment. La mémoire va se conserver de message en message en
empilant les messages dans une liste de dictionnaire. Le message initial est:

```python
messages = [​
    {​
        'role': 'system',​
        'content': 'Tu es un assistant technique professionnel. Tu dois aider le client en te basant sur les informations fournies dans la discussion.'
    }
]​
```

On empile les messages de l’utilisateur

```python
messages.append({'role': 'user', 'content': user_input})
```

Mais aussi les messages de l’assistant

```python
messages.append({'role': 'assistant', 'content': assistant_reply})
```

La fonction d’ollama `ollama.generate` prend une chaîne de caractère en entrée et non un
dictionnaire, il serait toutefois possible de lui donner une très longue chaîne de caractère en
précisant les rôles dans laquelle on ajouterait manuellement les rôles pour savoir qui à dit
quoi, c’est ce qu’on faisait autrefois (préhistoire).
Pour la gestion d’un dictionnaire il y a la fonction `ollama.chat`


````python
def chatbot_memory():
    # 1. contexte initialisation
    messages = [
        {
            "role": "system",
            "content": "tu es un assitant technique professionnel. Tu dois aider le client en ce bassant sur te basant sur les informations fournies dans la discussion.",
        }
    ]

    while True:
        user_input = input("\nClient : ")
        if user_input.lower() in ["exit", "stop", "fin"]:
            break
        # 2. Stack Question Client
        messages.append({"role": "user", "content": user_input})

        # 3. .chat gère listes de dictionnaires
        reponse = ollama.chat(model="llama3.2:3b", messages=messages)
        assistant_reply = reponse["message"]["content"]

        print("IA :", assistant_reply)

        # 4. Stack Response IA
        messages.append({"role": "assistant", "content": assistant_reply})


chatbot_memory()
````

## CHATBOT LANG CHAIN


#### Concepts introduits :

- **HumanMessage & AIMessage :** Ce sont des classes spécifiques qui typent l'origine du message. Utiliser ces objets permet à LangChain de formater automatiquement la discussion selon le modèle choisi (Ollama, OpenAI, etc.), garantissant une portabilité totale de votre code.
- **ChatPromptTemplate :** Contrairement à une simple chaîne de caractères, ce template permet de définir une structure rigide et réutilisable pour vos prompts,facilitant la gestion des consignes système et des entrées utilisateurs.
- **MessagesPlaceholder :** C'est l'outil clé pour la gestion de la mémoire. Il agit comme un "réservoir" dynamique à l'intérieur de votre prompt. Il indique à LangChain exactement où injecter l'historique des échanges accumulés, sans avoir à manipuler manuellement des listes ou des concaténations complexes.

Le message de départ devient:

```python
prompt = ChatPromptTemplate.from_messages([​
    ("system", "Tu es un assistant technique pro. Réponds de manière concise."),​
    MessagesPlaceholder(variable_name="chat_history"),​
    ("human", "{input}"),​
])
```

En utilisant ces objets, vous ne construisez plus un script pour un modèle précis, vous construisez une architecture de communication capable de piloter n'importe quelle IA du marché.

### La chaîne, le concept de LCEL:

Le Pipe, opérateur “|” (barre verticale et non un “l” ou “i” majuscule ) est le cœur de la syntaxe moderne de LangChain, appelée LCEL (LangChain Expression Language). Au lieu d'écrire des fonctions imbriquées complexes comme model(`prompt(input)`), on utilise le pipe pour créer une séquence logique où la sortie de l'un devient l'entrée du suivant.

Pour l’exercice on utilisera la chaîne suivante :

```python
chain = prompt | llm
```

​
où (à importer)

```python
llm = ChatOllama(model="llama3.2:3b", temperature=0)
```

Cela indique que le prompt rentre ensuite dans le llm. C’est un outils complexe qui permet de gérer des prompts par batch (`chain.batch([q1, q2, ...])`) ou d’utiliser des modèles de secours (`chain = (prompt | llm).with_fallbacks([backup_llm])`) et bien d’autres choses.


````python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage
from langchain_ollama import OllamaEmbeddings

# from langchain_chroma import Chroma
from langchain_community.vectorstores import PGVector

llm = ChatOllama(model="llama3.2:3b", temperature=0)

# On définit l'outil de vectorisation (le traducteur Texte -> Mathématiques)
embeddings = OllamaEmbeddings(model="nomic-embed-text")

# La connexion au conteneur Docker PostgreSQL
CONNECTION_STRING = "postgresql+psycopg2://postgres:password@localhost:5432/postgres"

# # On crée la base de données sur ton disque dur (dans le dossier ./ma_memoire_profonde)
# vectorstore = Chroma(
#     collection_name="historique_global",
#     embedding_function=embeddings,
#     persist_directory="./ma_memoire_profonde",
# )

vectorstore = PGVector(
    connection_string=CONNECTION_STRING,
    collection_name="historique_assistant",
    embedding_function=embeddings,
    use_jsonb=True,  # Optimisation vitale pour PostgreSQL
)

prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "Tu es un assistant technique pro. Réponds de manière concise.\n"
            "Voici des archives de nos précédentes conversations qui peuvent t'aider :\n"
            "{archives}",
        ),
        MessagesPlaceholder(variable_name="chat_history"),
        ("human", "{input}"),
    ]
)

chain = prompt | llm

history = []
k = 4  # Fenêtre glissante pour la VRAM

print("Assistant initialisé. (Tapez 'exit' pour quitter)")

while True:
    user_input = input("\nClient : ")
    if user_input.lower() in ["exit", "stop", "fin"]:
        break

    # Optimisation de la mémoire vive (Sliding Window)
    history_tronque = history[-k:] if len(history) > k else history

    # On cherche les 3 morceaux les plus proches sémantiquement de la question
    docs_du_passe = vectorstore.similarity_search(user_input, k=3)
    context_archives = "\n".join([doc.page_content for doc in docs_du_passe])

    #  Invocation de l'IA avec les 3 paramètres
    try:
        response = chain.invoke(
            {
                "input": user_input,
                "chat_history": history_tronque,
                "archives": context_archives,
            }
        )

        # Affichage (response est un objet AIMessage, on extrait le texte avec .content)
        response_clean = response.content.strip()
        print("IA :", response_clean)

        # Mise à jour de la mémoire à COURT terme (RAM)
        history.append(HumanMessage(content=user_input))
        history.append(AIMessage(content=response_clean))

        echange_formate = f"Client : {user_input} | IA : {response_clean}"
        vectorstore.add_texts(texts=[echange_formate])

    except Exception as e:
        print(f"Error: {e}")
````

````python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Ingestion : Chagement du document texte
loader = TextLoader("./docs_cours/menu.txt", encoding="utf-8")
documents = loader.load()

# Chunking : Découpage intelligent (Recursive)
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300, chunk_overlap=50, separators=["\n\n", "\n", ".", " "]
)
chunks = text_splitter.split_documents(documents)

# Vérification
print(f"Document découpé en {len(chunks)} morceaux (chunks).")
print("--- Voici le chunk n°1 ---")
print(chunks[0].page_content)
````

````python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage
from langchain_ollama import OllamaEmbeddings
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langchain_chroma import Chroma

# Models
llm = ChatOllama(model="llama3.2:3b", temperature=0)
embeddings = OllamaEmbeddings(model="nomic-embed-text")

# Création de la base Chroma à partir de tes chunks
vectorstore = Chroma.from_documents(documents=chunks, embedding=embeddings)

# Le Retriever : On paramètre le moteur de recherche pour renvoyer les 3 meilleurs résultats
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})

# Le Prompt : Plus stricte pour moins d'hallucination.
template = """Tu es un serveur de restaurant virtuel. Réponds à la question du client en utilisant UNIQUEMENT les informations du menu ci-dessous.
Si la réponse n'y est pas, dis poliment que tu ne sais pas.

Menu (Contexte) :
{context}

Question client : {question}"""

prompt = ChatPromptTemplate.from_template(template)


# Fonction utilitaire : Assemble les 3 morceaux trouvés en un seul texte lisible
def format_docs(docs):
    return "\n\n".join([d.page_content for d in docs])


# 3. L'architecture LCEL (Le Pipe RAG)
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)

while True:
    user_input = input("\nClient : ")
    if user_input.lower() in ["exit", "stop", "fin"]:
        break

    #  Invocation de l'IA avec les 3 paramètres
    try:
        reponse = rag_chain.invoke(user_input)
        print(f"IA Serveur : {reponse}")

    except Exception as e:
        print(f"Error: {e}")
````
